```python
from selkie.wap.ui import Element
```

```python
class JSStyle:
    pass
```

```python
class JSNode:

    def __init__ (self, label, *args):
        self.label = label
        self.args = args
        self.style = JSStyle()
        self.children = []

    def appendChild (self, child):
        self.children.append(child)

    def __repr__ (self):
        return f'<JSNode {self.label}>'
```

```python
class TestDocument (Element):

    def __init__ (self):
        self.document = self
        self._js = JSNode('document')

    def createTextNode (self, text):
        return JSNode('#TEXT', text)

    def createElement (self, label):
        return JSNode(label)

    def __repr__ (self):
        return '<TestDocument>'
```

```python
document = TestDocument()
```
